# Getting Started: Interactive Serverless Spark Notebooks

Use this notebook template to connect a local Python kernel to **Managed Service for Apache Spark** using the `google-cloud-spark-connect` client library (`ManagedSparkSession`).

### Steps to Execute Interactively

1. **Create a Runtime Profile (Optional, first-time setup)**:
   In the Activity Bar, open **Data Agent Kit > Apache Spark > Serverless > New Runtime Profile** to configure reusable Serverless Spark settings (or skip to use default Serverless Spark settings).

1. **Select a Local Python Kernel**:
   Click **Select Kernel** in the top-right corner of this notebook and choose a local Python environment (Python 3.12+).
3. **Configure & Run**:
   Set your `PROJECT_ID`, `REGION`, and `RUNTIME_PROFILE` in **Step 2** below and execute the cells in order.

> **Tip:** Provisioning a new Serverless Spark session might take **~1–2 minutes** on the first run. Once active, subsequent cells reuse the live session and run in seconds.

## 1. Install the Managed Spark Connect Client
Ensure `google-cloud-spark-connect` is installed in your local Python kernel environment.

```python
# Uncomment and run if google-cloud-spark-connect is not yet installed in your local kernel:
# %pip install -q -U 'google-cloud-spark-connect[interactive,client]'
```

## 2. Initialize an Interactive Serverless Spark Session
Create a `ManagedSparkSession` using either default Serverless Spark settings or a **Runtime Profile** (Session Template) created in Data Agent Kit.

```python
from google.cloud.dataproc_v1 import Session
from google.cloud.managed_spark_connect import ManagedSparkSession

# 1. Set your Google Cloud project and region
PROJECT_ID = 'YOUR-PROJECT-ID'
REGION = 'GOOGLE_CLOUD_REGION'

# 2. (Optional) Attach a Runtime Profile created in Data Agent Kit > Apache Spark > Serverless
RUNTIME_PROFILE = 'YOUR-RUNTIME-PROFILE'

session_config = Session()
session_config.session_template = (
    f'projects/{PROJECT_ID}/locations/{REGION}/sessionTemplates/{RUNTIME_PROFILE}'
)

# 3. Create or connect to the interactive Serverless Spark session
# Tip: Remove .sessionConfig(session_config) to use default Serverless Spark settings
spark = (
    ManagedSparkSession.builder
    .projectId(PROJECT_ID)
    .location(REGION)
    .sessionConfig(session_config)
    .getOrCreate()
)
```

## 3. Write your Spark code
Once your `spark` session is active, you can load tables and run PySpark DataFrame transformations interactively.

```python
# Load the table into a Spark DataFrame
df = (
    spark.read.format("iceberg")
    .load("`lakehouse-public-data`.chicago_taxi_trips.taxi_trips")
    .limit(100)
)

df.show(10, truncate=False)
```
